# Database Updates

Two functions update the database `db.update()` and `db.add_experiment(rel_path=...)`.

The latter adds a single experiment folder, parsing the relevant data/metadata. It add events from the `Events.csv` files (if they exist), checks H5 files, and gets TIFFs metadata. It also performs some checks to see if the data format is as expected.

The former searches the whole `ImagingData` folder (i.e. the DB `main_folder`) for TIFFs inside a `raw` folder. Then, the folder that contains the `raw` folder is the "experiment folder" which will be given to a `db.add_experiment(...)`.

## Manual Edits

Note that modifying these `DataFrames` will not modify the database.

```python
db.experiments.loc[10, "exp_name"] = ""
```

```python
db.experiments
```

The commands above change the data in 'db', but not in the actual database.

To erase those changes and get the original data back, run the following command.

```python
db._reset_caches()
```
