## Connecting to the DB

To connect to the database, you have to create an instance of the `Database` class.

```python
# Change the path below to the `ImagingData` path in your computer
#
# If you are on a MacOS, just paste the path from Finder
#   (you can remove the `r` in front of the string).
#
# If you are using Windows, you will have to use r"PATH" for all your paths.
#   (This is because "\" has another meaning in Python strings, so to use it
#    you have to tell Python that this is a raw string. )

db = Database(r"tmp\test_server")
```

# Analysis/Processing Groups

Groups are sets of experiments that you want to analyze/process together (an experiment can be in many groups). They are used to record what processes were run on which experiments and what errors you got along the way. By default `odyn` creates one group for each experiment.

```python
# This command will give you a list of all groups currently in the database
#
# Note that Group 1 is actually the 0th element of the list
#   (Python is 0-indexed but SQL is 1-indexed so this is the default, unfortunately)

db.groups
```

```python
# Getting the Group with group_id=151, for example
group = db.groups[150]

# Python prints the output of the last line in a cell by default
# So, the next line will show you some data about Group 151
group
```

```python
# This next function will run motion correction on the group as a whole.
# Since there is only one experiment in this group, it will only run
# motion correction on that experiment folder files.
group.run_motion_correction()

# You can hover over run_motion_correction to get help for that function
# Or you if want to see what are all the Group functions, you can hover
# over `Group` in the next line or you can write a dot next to it, to
# get the autocomplete suggestions`
Group
```
